```python
import opensmile
import time

smile = opensmile.Smile(
    feature_set=opensmile.FeatureSet.eGeMAPSv02,
    feature_level=opensmile.FeatureLevel.Functionals,
)

start = time.time()
features = smile.process_file('test.wav')
elapsed = time.time() - start

# Extract the 6 actionable speech therapy metrics
row = features.iloc[0]

print("=== Speech Health Profile ===\n")

# 1. F0 (pitch) — prosody / monotone detection
print(f"F0 Mean:          {row['F0semitoneFrom27.5Hz_sma3nz_amean']:.2f} semitones")
print(f"F0 Std Dev:       {row['F0semitoneFrom27.5Hz_sma3nz_stddevNorm']:.3f} (normalized)")
print(f"  → {'Monotone' if row['F0semitoneFrom27.5Hz_sma3nz_stddevNorm'] < 0.1 else 'Normal intonation'}")

# 2. Jitter — vocal fold irregularity
print(f"\nJitter:           {row['jitterLocal_sma3nz_amean']:.4f}")
print(f"  → {'Elevated' if row['jitterLocal_sma3nz_amean'] > 0.02 else 'Normal'} (threshold ~0.02)")

# 3. Shimmer — amplitude perturbation
print(f"\nShimmer:          {row['shimmerLocaldB_sma3nz_amean']:.3f} dB")
print(f"  → {'Elevated' if row['shimmerLocaldB_sma3nz_amean'] > 0.5 else 'Normal'} (threshold ~0.5 dB)")

# 4. HNR — breathiness
print(f"\nHNR:              {row['HNRdBACF_sma3nz_amean']:.2f} dB")
print(f"  → {'Breathy' if row['HNRdBACF_sma3nz_amean'] < 10 else 'Normal'} (threshold ~10 dB)")

# 5. Loudness — breath support / volume control
print(f"\nLoudness Mean:    {row['loudness_sma3_amean']:.3f}")
print(f"Loudness Std Dev: {row['loudness_sma3_stddevNorm']:.3f} (normalized)")

# 6. Speech rate / pausing
print(f"\nSpeaking Rate:    {row['loudnessPeaksPerSec']:.2f} syllables/sec")
print(f"Voiced Segments:  {row['VoicedSegmentsPerSec']:.2f} /sec")
print(f"Mean Pause Len:   {row['MeanUnvoicedSegmentLength']:.3f} s")
print(f"Mean Voiced Len:  {row['MeanVoicedSegmentLengthSec']:.3f} s")

print(f"\n  Processing time: {elapsed:.3f}s")
```
